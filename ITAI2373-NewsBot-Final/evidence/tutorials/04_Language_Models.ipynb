```python
from pathlib import Path
import os,sys,subprocess
root=Path.cwd()
if not (root/'newsbot2').is_dir():
    options=[root/'final-newsbot-2.0',root.parent]
    root=next((p for p in options if (p/'newsbot2').is_dir()),None)
    if root is None:
        root=Path.cwd()/'NewsBot_Project'
        subprocess.run(['git','clone','--depth','1','https://github.com/jtuozzo1616/Final-Jose-Group-NewsBot-Intelligence-System.git',str(root)],check=True)
        root=root/'final-newsbot-2.0'
        subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(root/'requirements-dev.txt')],check=True)
os.chdir(root);sys.path.insert(0,str(root))
from newsbot2.io import import_articles
from newsbot2.engine import NewsBotEngine
articles=import_articles(root/'data/demo_news.jsonl')
engine=NewsBotEngine(articles)
```

```python
from newsbot2.advanced import expand_query
print(engine.analyze('demo-tech-1')['summary'])
print(expand_query('robot'))
print(engine.search('robot navigation sensors'))
```
